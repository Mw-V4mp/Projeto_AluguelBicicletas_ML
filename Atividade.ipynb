```python
from sklearn.linear_model import LinearRegression

# Duração do aluguel (minutos)
x = [[10],[15],[20],[25],[30],[35],[40],[50],[55],[60]]
# Preço pago (R$) 
y = [10,13,14,18,19,23,24,30,32,34]

modelo = LinearRegression()

# Treinando o modelo
modelo.fit(x,y)

predicao = modelo.predict([[45]])

print(f"Predição de lucro: R$ {predicao[0]:,.2f}")
```
